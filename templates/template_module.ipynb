This Jupyter Notebook contains examples of how we structure the different modules.
We want all training modules to follow a consistent structure as much as possible.
Please use the patterns described below.

We always follow the same structure: first explain the theory or concept, then show example code, and finally provide exercise(s) for the participant to practice the concept.

#### Theory

Make the theory as clear as possible. Videos and images are welcome and can be used freely. It is also important to reference the documentation of relevant packages/libraries.

Try not to cover too much theory at once. Keep it interactive with frequent alternation between explanation and practice.

#### Example Code

Example code does not need to be executed by the participants and can be embedded in markdown. We want participants to type code themselves as much as possible — this is how you learn the most while programming.

Code examples should be small and clear.

```python
### CODE EXAMPLE
import requests

url = "example.com/api/data"
requests.get(url)
```

#### Answers Notebook Pattern

In the Answers notebook, each solution code cell should be **preceded** by a markdown explanation cell. This cell should contain:
- A brief explanation of the approach (2-4 sentences)
- The key concept or library feature being used
- Common pitfalls or alternatives where relevant

Example:

> **Explanation:** We use `requests.get()` to perform an HTTP GET request to the API endpoint. The response object contains the status code and body. A common pitfall is forgetting to check the status code before accessing the response data.
